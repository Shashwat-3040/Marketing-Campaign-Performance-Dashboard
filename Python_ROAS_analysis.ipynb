```python
import pandas as pd

df = pd.read_csv('global_ads_performance_dataset.csv')
df['CTR'] = df['clicks'] / df['impressions']
df['CPC'] = df['ad_spend'] / df['clicks']

platform_summary = df.groupby('platform')[['CTR', 'CPC', 'CPA', 'ROAS']].mean()
print(platform_summary)

platform_summary_sorted = platform_summary.sort_values('ROAS', ascending=False)
print(platform_summary_sorted)

best_roas = platform_summary_sorted['ROAS'].iloc[0]
worst_roas = platform_summary_sorted['ROAS'].iloc[-1]
multiplier = best_roas / worst_roas
print(f"Best platform has {multiplier:.1f}x higher ROAS than worst platform")
```

```python
total_impressions = df['impressions'].sum()
total_clicks = df['clicks'].sum()
total_conversions = df['conversions'].sum()

click_dropoff = 1 - (total_clicks / total_impressions)
conversion_dropoff = 1 - (total_conversions / total_clicks)

print(f"Drop-off from impression to click: {click_dropoff:.1%}")
print(f"Drop-off from click to conversion: {conversion_dropoff:.1%}")
```

```python
import pandas as pd

df = pd.read_csv('global_ads_performance_dataset.csv')
df['CTR'] = df['clicks'] / df['impressions']
df['CPC'] = df['ad_spend'] / df['clicks']
df['CVR'] = df['conversions'] / df['clicks']

platform_campaign = df.groupby(['platform', 'campaign_type'])['ROAS'].mean().sort_values(ascending=False)
print(platform_campaign)
```
